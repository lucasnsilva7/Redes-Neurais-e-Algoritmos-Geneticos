### Introdução 

Para resolver esse problema temos alguns caminhos que podemos seguir. O primeiro é garantir que em nenhum momento o algoritimo gere indivíduos inválidos, isso é, que não possuem pelo menos uma vogal. A segunda é penalizar um indivíduo na função objetivo se ele não possuir pelo menos uma vogal. No caso, usei essa segunda opção pela simplicidade dela.

Realizei uma alteração na função de criação de indivíduos e da população, mas essa alteração apenas mudou o nome delas. 

A alteração importante no algoritmo foi realizada na função objetivo. Basicamente foi necessário definir quais são as vogais e criar uma variável booleana chamada `possui_vogal` que inicialmente começa como `False`, se houver pelo menos uma vogal no indivíduo essa variável fica definida como `True`. 

Essa parte é importante porque se `possui_vogal` for `False` a distância é definida como 0.01, penalizando o indivíduo. Se for `True` um for é percorrido pegando cada letra do candidato e a sua complementar, que é pera percorrendo o candidato de trás para frente. Exemplo: se a palavra for `nivel` o for vai pegar a letra como sendo n e o complementar como sendo l, após, ele vai pegar a letra i e o complementar e, por fim, ambos pegarão a letra v. 

Para evitar que o resto da lista fosse percorrido, o que seria inútil, eu defini o centro da minha palavra como sendo a aproximação para mais do len da lista candidato dividido por dois a partir da função `math.ceil`, dessa forma, para palavras com len par ele para realmente em len/2 e para len impar ele para em len/2 + 0.5, garantindo que ambas as letras sejam observadas. Eu dou um `break` no for se essa letra do centro já tiver sido observada.

Após pegar essas letras eu somo na distância o absoluto no número que representa a letra (pela função `ord()`) pelo número que representa o complementar. 

### Importando bibliotecas

```python
from string import ascii_lowercase

from funcoes import populacao_palindromo as cria_populacao
from funcoes import funcao_objetivo_pop_palindromo as funcao_objetivo
from funcoes import selecao_torneio_min as funcao_selecao
from funcoes import cruzamento_uniforme as funcao_cruzamento
from funcoes import mutacao_simples as funcao_mutacao1
from funcoes import mutacao_salto as funcao_mutacao2
```

### Definindo constantes 

```python
CARACTERES_POSSIVEIS = ascii_lowercase

TAMANHO_PALAVRA = 5
TAMANHO_POPULACAO = 100
CHANCE_DE_CRUZAMENTO = 0.5
CHANCE_DE_MUTACAO = 0.025
TAMANHO_TORNEIO = 3
NUM_GERACOES = 500
NUM_PALINDROMOS = 10
```

### Encontrando os palindromos

Observe que o código abaixo funciona evoluindo o algoritmo `NUM_PALINDROMOS` vezes, no caso 10, para encontrar os 10 palindromos. O código cria uma população e a partir dela entra em um laço até achar um indivíduo com um bom fitness. Assim, ele printa esse indivíduo e entra no segundo valor de `palindromo` no for para criar uma nova palavra que é um palindromo.

```python
for palindromo in range(NUM_PALINDROMOS):
    
    # Criar população
    populacao = cria_populacao(TAMANHO_POPULACAO, TAMANHO_PALAVRA, CARACTERES_POSSIVEIS)
    
    menor_fitness_geral = float("inf")
    geracao = 0 

    while menor_fitness_geral != 0:

        # Seleção
        fitness = funcao_objetivo(populacao)        
        selecionados = funcao_selecao(populacao, fitness, TAMANHO_TORNEIO)
        
        # Cruzamento
        proxima_geracao = []
        for pai, mae in zip(selecionados[::2], selecionados[1::2]):
            individuo1, individuo2 = funcao_cruzamento(pai, mae, CHANCE_DE_CRUZAMENTO)
            proxima_geracao.append(individuo1)
            proxima_geracao.append(individuo2)
        
        # Mutação
        funcao_mutacao1(proxima_geracao, CHANCE_DE_MUTACAO, list(CARACTERES_POSSIVEIS))
        funcao_mutacao2(proxima_geracao, CHANCE_DE_MUTACAO, list(CARACTERES_POSSIVEIS))
        
        # Encerramento
        populacao = proxima_geracao
        geracao += 1
        
        fitness = funcao_objetivo(populacao)
        menor_fitness_observado = min(fitness)
        
        if menor_fitness_observado == 0:
            menor_fitness_geral = menor_fitness_observado
            indice = fitness.index(menor_fitness_observado)
            candidato = populacao[indice]
            palavra = "".join(candidato)
            print(f"Palindromo {palindromo + 1}: {palavra}, encontrado na geração {geracao}")
```
